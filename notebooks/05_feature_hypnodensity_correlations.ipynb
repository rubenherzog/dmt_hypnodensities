# 05 · Hypnodensity–feature correlations

This notebook is a focused derivative of `03_feature_associations.ipynb`. It reuses the
same validated feature/hypnodensity assembly, but replaces the treatment-effect pipeline
with a direct, descriptive correlation analysis between raw epoch-level EEG features and
raw epoch-level hypnodensity probabilities.

The analysis unit throughout is the **epoch** (one row per recording × block × epoch ×
electrode × stager, after joining monoelectrode features onto the aligned staging output).
Four correlation analyses are run, each pooling data differently:

1. **Global** — one feature × probability correlation matrix per stager, pooled across all
   subjects, epochs, electrodes, experimental phases and doses. This looks for a
   feature–probability mapping that holds on average across the whole dataset.
2. **Per electrode** — the same matrix, repeated separately for every electrode. This asks
   whether the global mapping found in (1) actually depends on which electrode is used.
3. **Per phase × dose segment** — the same global matrix, computed separately within each
   of the four `experimental_label × condition` segments (before-placebo, before-DMT,
   after-placebo, after-DMT). This asks whether the feature–probability mapping itself
   shifts depending on where in the experiment (and under which drug condition) it is
   estimated — i.e. whether the statistical model implicitly partitions its input space.
4. **Per electrode, per phase × dose segment** — analysis (3) repeated separately for every
   electrode, to check whether that partitioning itself is electrode-dependent.

Spearman correlation is used throughout: hypnodensity probabilities are bounded and often
far from normally distributed, and several EEG features are heavy-tailed, so a rank-based,
monotonic measure is a safer default than Pearson's r.

## i) Imports

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import display

from dmt_hypnodensities import (
    assemble_outputs, feature_value_columns, join_epoch_features_hypnodensities,
    load_config, save_figure, save_table, stage_feature_effect_correlations,
)
```

## ii) Parameters and persisted run

`RUN_NAME` points at the same precomputed feature/hypnodensity outputs used by
`03_feature_associations.ipynb`; nothing is recomputed here. `PHASES` and `DOSES` define
the four-way segmentation used by analyses 3 and 4 (the `late` and `outside` experimental
labels are excluded — only the paired before/after window is used).

Unlike `03_feature_associations.ipynb`, this notebook resolves the run's `recordings` /
`tables` / `figures` directories directly under `output_dir`, instead of going through
`prepare_run`. `prepare_run` is designed to guard a *pipeline* run (it refuses to reuse a
run directory if today's `configs/analysis.yaml` doesn't hash-match the config that was
used to compute it — e.g. the gap-tolerance sweep runs were each produced with a different
`segmentation.gap_tolerance_seconds`, so only one of them can ever match the file's current
value). This notebook only *reads* already-computed outputs and writes derived tables and
figures, so that provenance guard is unnecessary friction here — it would otherwise force
editing `analysis.yaml` back and forth just to switch which precomputed run you're looking
at.

```python
CONFIG_PATH = next(path.resolve() for path in (Path('configs/analysis.yaml'), Path('../configs/analysis.yaml')) if path.is_file())
RUN_NAME = 'gap_sensitivity_000s_gssc_yasa_sleepfm_cpu_v1'
STAGES = ('W', 'N1', 'R')
PROBABILITIES = tuple(f'prob_{stage}' for stage in STAGES)
STAGERS = ('gssc',)
PHASES = ('before', 'after')
DOSES = ('placebo', 'DMT')
CORRELATION_METHOD = 'spearman'
MIN_OBSERVATIONS = 30
TOP_N_RANKED_FEATURES = 10

RUN_ROOT = load_config(CONFIG_PATH).output_dir / RUN_NAME
RECORDINGS_DIR = RUN_ROOT / 'recordings'
TABLES_DIR = RUN_ROOT / 'tables'
FIGURES_DIR = RUN_ROOT / 'figures'
TABLES_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

tables = assemble_outputs(RECORDINGS_DIR, strict=True)
FEATURES = feature_value_columns(tables.features)
display({'n_features': len(FEATURES), 'features': FEATURES})
```

## iii) Epoch-level analytical table

Monoelectrode features are joined onto the aligned staging output on the exact
recording–block–epoch–electrode key, matching `03_feature_associations.ipynb`. Each row is
one epoch for one electrode and one stager, carrying both the five hypnodensity
probabilities and every extracted feature.

```python
epoch_analysis = join_epoch_features_hypnodensities(
    tables.features, tables.hypnodensities, strict=True
)
epoch_analysis = epoch_analysis.loc[epoch_analysis['stager'].isin(STAGERS)].copy()
save_table(epoch_analysis, TABLES_DIR / 'epoch_feature_hypnodensity.parquet')
display(epoch_analysis.shape, epoch_analysis.groupby('stager').size())
```

## iv) Plotting helpers

`stage_feature_effect_correlations` is generic enough to correlate any two sets of columns
within an arbitrary grouping — it is reused here for raw values instead of treatment
effects by passing the probability and feature columns directly, with the grouping columns
that define each analysis below (`stager` alone for the global analysis, plus `electrode`
and/or `experimental_label`/`condition` for the more granular ones).

Two small local helpers turn one such correlation table into a stage × feature matrix, and
lay out several matrices side by side for comparison. They are kept in the notebook because
each analysis below needs a different combination of panels.

```python
def pivot_correlation_matrix(table, row='stage', col='feature', value='correlation'):
    """Reshape one long correlation table into a stage x feature matrix."""

    matrix = table.pivot(index=row, columns=col, values=value)
    order = [stage for stage in STAGES if stage in matrix.index]
    return matrix.reindex(order)


def plot_correlation_grid(
    panels, ncols=3, value_label='Correlation r', vmin=-1, vmax=1,
    cmap='coolwarm', figsize=None
):
    """Plot a dict of {panel title: stage x feature matrix} as a shared-scale grid."""

    n_panels = len(panels)
    ncols = min(ncols, n_panels)
    nrows = int(np.ceil(n_panels / ncols))
    figure, axes = plt.subplots(
        nrows, ncols, figsize=figsize or (4 * ncols, 10 * nrows), squeeze=False,
        constrained_layout=True,
    )
    image = None
    for index, (axis, (title, matrix)) in enumerate(zip(axes.flat, panels.items())):
        matrix = matrix.T
        image = axis.imshow(matrix.to_numpy(), vmin=vmin, vmax=vmax, cmap=cmap, aspect='auto')
        axis.set_xticks(np.arange(matrix.shape[1]), matrix.columns, fontsize=9)
        if index % ncols == 0:
            axis.set_yticks(np.arange(matrix.shape[0]), matrix.index, fontsize=7)
        else:
            axis.set_yticks([])
        axis.set_title(title)
    for axis in axes.flat[n_panels:]:
        axis.axis('off')
    if image is not None:
        figure.colorbar(image, ax=axes, label=value_label, shrink=0.8)

    return figure, axes
```

## v) Analysis 1 — Global feature × probability correlations

One correlation matrix per stager, pooling every subject, epoch, electrode, experimental
phase and dose together. This is the coarsest possible look at the data: which features
track which hypnodensity probabilities on average, independent of where or when the epoch
was recorded.

```python
global_correlations = stage_feature_effect_correlations(
    epoch_analysis,
    stage_columns=PROBABILITIES,
    feature_columns=FEATURES,
    group_by=('stager',),
    methods=(CORRELATION_METHOD,),
    min_observations=MIN_OBSERVATIONS,
)
save_table(global_correlations, TABLES_DIR / 'global_feature_probability_correlations.csv')
display(global_correlations.head())
```

```python
pooled_correlations = global_correlations.loc[global_correlations['stager'].eq(DEFAULT_STAGER)]
pooled_matrix = pivot_correlation_matrix(pooled_correlations)
feature_order = pooled_matrix.loc['W'].sort_values(ascending=False).index
pooled_matrix = pooled_matrix[feature_order]

figure, axes = plot_correlation_grid(
    {'All electrodes pooled, features ranked by W': pooled_matrix},
    ncols=1, figsize=(6, 8),
)
save_figure(figure, FIGURES_DIR / f'global_feature_probability_correlations_ranked_by_W_{DEFAULT_STAGER}')
plt.show()
```

Follows the logic of the four-panel figure in `02_hypnodensity_analysis.ipynb` — violin
distributions split by condition, with each subject’s two values connected by a line —
but applied to raw EEG feature values instead of hypnodensity probabilities, and
without its fourth LME forest panel. The `TOP_N_CONDITION_FEATURES` features most
positively and most negatively correlated with `prob_W` (the same `feature_order`
ranking used in analysis 6) are shown together, one panel per phase, split by condition.
Subject-level phase means pool across epochs and electrodes, as in analysis 1. Because
the ten features live on very different natural scales, each one is z-scored across all
pooled subject × phase × condition observations purely for display — only relative,
within-feature shifts between conditions are meaningful here, not the absolute values.
